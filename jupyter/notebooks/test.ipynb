```python
import chromadb
from chromadb.config import Settings

# Connect to the ChromaDB HTTP server
client = chromadb.HttpClient(host="chromadb", port=8000)

# Create or get an existing collection
collection = client.get_or_create_collection("jupyter_test")

# Insert sample documents
documents = [
    "Jupyter integrates with ChromaDB.",
    "This is a test document.",
    "ChromaDB is a vector database.",
    "Streamlit can also connect to ChromaDB."
]
ids = ["doc1", "doc2", "doc3", "doc4"]

# Optional: delete existing records (to avoid duplication if rerunning)
try:
    collection.delete(ids=ids)
except:
    pass

collection.add(documents=documents, ids=ids)

# Run a query
query_result = collection.query(
    query_texts=["test chromadb from jupyter"],
    n_results=2
)
print(query_result)
```
